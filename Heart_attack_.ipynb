```python
import numpy as np
import pandas as pd

```

```python
data = pd.read_csv('/root/.cache/kagglehub/datasets/fatemehmohammadinia/heart-attack-dataset-tarik-a-rashid/versions/1/Medicaldataset.csv')
```

```python
X = data.drop('Result', axis=1)
y = data['Result']
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y_encoded = le.fit_transform(y)
print(f"Original classes: {le.classes_}")
print(f"Encoded values: {le.transform(le.classes_)}")
print(f"First 5 encoded target values: {y_encoded[:5]}")
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
print("Unique values in Gender column:", X_train['Gender'].unique())

# Based on the unique values, 'Gender' (0 and 1) is a categorical feature. All other features are numerical.
categorical_features = ['Gender']
numerical_features = X_train.columns.drop(categorical_features).tolist()
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Create a column transformer for preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ])

# Apply the preprocessing to the training and testing sets
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print(f"X_train_processed shape: {X_train_processed.shape}")
print(f"X_test_processed shape: {X_test_processed.shape}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Initialize and train the Logistic Regression model
model = LogisticRegression(random_state=42)
model.fit(X_train_processed, y_train)

# Make predictions on the preprocessed test set
y_pred = model.predict(X_test_processed)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy:.4f}")
```
